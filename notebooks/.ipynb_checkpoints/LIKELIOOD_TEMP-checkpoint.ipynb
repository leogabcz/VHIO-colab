# NB trajectory pipeline — methods walkthrough
## Section 2. The negative binomial null model and its dispersion

The pipeline tests each patient clone against a **null model**: what its counts would
look like if its frequency stayed constant and only measurement noise acted. This
section defines that model, shows how its single noise parameter, the dispersion
$\phi$, is estimated from healthy donors, checks the estimator by simulation, and
builds the reference $\phi$ used for patients.

| | Question | Outcome |
|---|---|---|
| 2.1 | What is the model? | NB with offsets $L_t$ and dispersion $\phi$ |
| 2.2 | Is a Poisson enough? | No: the noise is far larger than read sampling |
| 2.3 | How is $\phi$ estimated? | Truncated likelihood, profiled over clone frequencies, Cox–Reid adjusted |
| 2.4 | Does the estimator recover a known $\phi$? | Simulation study on the donors' real designs |
| 2.5 | Is one $\phi$ enough for all clones? | No: $\phi$ rises with clone frequency, as in DESeq2/edgeR |
| 2.6 | Which $\phi$ do patients get? | The noisiest donor at each frequency, $\phi_\text{ref}(f)$ |
| 2.7 | Is the null calibrated? | Leave-one-donor-out: stability of $\phi_\text{ref}(f)$ and false-positive rates |

Section 1 (exploratory analysis) and the sample QC are assumed: the failed donor
sample (HDBR06 day 21) is removed here.

```python
import sys
sys.path.insert(0, '..')

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

from tcr_pipeline.normalization import tmm_normalize
from tcr_pipeline.data_io import sample_qc, drop_failed_samples
from tcr_pipeline import nb_glm as nb

warnings.filterwarnings('ignore')
pd.set_option('display.width', 160)

DONOR_COLORS = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100', '#e87ba4', '#008300']
GREY, INK = '#8a8984', '#3d3c39'
plt.rcParams.update({
    'figure.dpi': 110, 'axes.spines.top': False, 'axes.spines.right': False,
    'axes.grid': True, 'grid.color': '#e6e5e0', 'grid.linewidth': 0.6,
    'axes.edgecolor': GREY, 'axes.labelcolor': INK, 'xtick.color': '#52514e',
    'ytick.color': '#52514e', 'lines.linewidth': 1.6, 'lines.markersize': 5,
    'legend.frameon': False, 'font.size': 9.5, 'axes.titlesize': 10})

N_JOBS = -1          # parallel workers for the trajectory test (2.7)

from matplotlib.ticker import FuncFormatter, LogLocator, NullFormatter

def tidy_log(fig):
    """Plain tick labels on log axes (2e-05, 1.5, 3 instead of 3x10^0)."""
    for ax in fig.axes:
        for axis, scale, subs in ((ax.xaxis, ax.get_xscale(), (1, 2, 5)),
                                  (ax.yaxis, ax.get_yscale(), (1, 1.5, 2, 3, 5, 7))):
            if scale == 'log':
                axis.set_major_locator(LogLocator(subs=subs))
                axis.set_major_formatter(FuncFormatter(lambda v, _: f'{v:g}'))
                axis.set_minor_formatter(NullFormatter())
```

```python
healthy_raw = pd.read_csv('../data/healthy_raw.csv', index_col=0)
d_qc = nb.attach_offset(tmm_normalize(drop_failed_samples(healthy_raw, sample_qc(healthy_raw))))

DONORS = sorted(d_qc['patient'].unique())
COLOR = dict(zip(DONORS, DONOR_COLORS))
short = lambda p: p.replace('_SEQTR', '').replace('_ReSeq', '')
print(len(d_qc), 'rows,', d_qc.groupby(['patient', 'day']).ngroups, 'samples,',
      len(DONORS), 'donors')
```

## 2.1 The model

For clone $c$ in sample $t$ of one person, the read count is

$$x_{ct} \sim \mathrm{NB}\!\left(\mu_{ct},\ \phi\right), \qquad
\mu_{ct} = L_t\,\lambda_c, \qquad
\mathrm{Var}(x_{ct}) = \mu_{ct} + \frac{\mu_{ct}^2}{\phi}.$$

* $\lambda_c$ is the clone's frequency. Under the **null** it is the same in every
  sample: the clone neither expands nor contracts.
* $L_t = D_t \times \text{TMM}_t$ is the sample's **offset**: its in-frame depth $D_t$
  (Section 1.1) times a TMM normalisation factor computed within the person, so that
  $\mu_{ct}$ is the expected count of a clone of frequency $\lambda_c$ in that sample.
* $\phi$ is the **dispersion**. The NB is a Poisson whose mean is itself multiplied by
  a random factor with mean 1 and squared coefficient of variation $1/\phi$: the
  sample-to-sample noise of library preparation, RNA content per cell, amplification.
  Small $\phi$ means noisy; $\phi \to \infty$ is a Poisson.

A patient clone is called non-neutral when its counts are too unlikely under this
model (Section 4). Everything therefore rests on the value of $\phi$: too large and
ordinary noise is called a change; too small and real changes are missed.

```python
offsets = (d_qc.groupby(['patient', 'day'])[['D_t', 'tmm_factor', 'lib']].first()
             .rename(columns={'lib': 'L_t'}))
offsets.index = offsets.index.set_levels(offsets.index.levels[0].map(short), level=0)
display(offsets.round({'D_t': 0, 'tmm_factor': 3, 'L_t': 0}))
```

## 2.2 Is a Poisson enough?

For each donor, take the clones detected in **every** sample, rescale their counts to a
common depth ($x_{ct}\,\bar L / L_t$), and compute each clone's mean and variance across
samples. A Poisson predicts variance = mean (grey dashed line).

```python
def mean_variance(g):
    days = np.sort(g['day'].unique())
    L = g.groupby('day')['lib'].first().reindex(days).to_numpy()
    X = (g.pivot_table(index='clono', columns='day', values='count', aggfunc='sum',
                       fill_value=0).reindex(columns=days, fill_value=0).to_numpy(float))
    Y = X[(X > 0).all(axis=1)] * (L.mean() / L)
    return Y.mean(axis=1), Y.var(axis=1, ddof=1)

fig, axes = plt.subplots(2, 3, figsize=(11, 6.2), sharex=True, sharey=True)
for ax, p in zip(axes.ravel(), DONORS):
    m, v = mean_variance(d_qc[d_qc['patient'] == p])
    edges = np.logspace(np.log10(m.min()), np.log10(m.max()), 22)
    k = np.digitize(m, edges)
    grp = [i for i in np.unique(k) if (k == i).sum() >= 20]
    mb = np.array([m[k == i].mean() for i in grp])
    vb = np.array([v[k == i].mean() for i in grp])
    ax.scatter(m, v, s=1, color=COLOR[p], alpha=0.08, rasterized=True)
    ax.plot(mb, vb, color=INK, lw=1.8, label='mean variance (binned)')
    mm = np.logspace(1, 5, 50)
    ax.plot(mm, mm, color=GREY, ls='--', lw=1.2, label='Poisson: Var = μ')
    ax.set(xscale='log', yscale='log', title=f'{short(p)} ({len(m):,} clones)')
for ax in axes[-1]:
    ax.set_xlabel('mean count (common depth)')
for ax in axes[:, 0]:
    ax.set_ylabel('variance across samples')
axes[0, 0].legend(fontsize=7.5, loc='upper left')
fig.tight_layout()
```

**What this shows.** The variance lies above the Poisson line for every donor, by up
to three orders of magnitude, and the gap widens with the mean: the noise has a component that scales
with $\mu^2$, which is what the NB's $\mu^2/\phi$ term describes. A Poisson model would
treat this ordinary sample-to-sample noise as real change and call most clones
non-neutral. (These clones are selected for being detected in every sample, which
trims the low tail at small means; the conclusion concerns the large means, where the
selection does not matter.)

The binned variance also grows somewhat **more slowly** than $\mu^2$ at the largest
means: abundant clones are relatively less noisy than rare ones. Section 2.5 returns to
this.

## 2.3 Estimating $\phi$

$\phi$ is estimated per donor from the donor's own clones, with three ingredients.

**1. Detected counts only, with a truncated likelihood.** A clone missing from a sample
is not a zero (Section 1): it may simply not have been sampled. Only detected counts
are used, and since a count is only reported when it reaches the sample's cutoff
$c_t$ (Section 1.2), each is modelled *conditional on detection*:

$$P(x_{ct} \mid x_{ct} \ge c_t) = \frac{\mathrm{NB}(x_{ct};\ \mu_{ct}, \phi)}{P_{NB}(X \ge c_t;\ \mu_{ct}, \phi)}.$$

Without the denominator, the counts that fell below the cutoff are silently missing
from the low tail and the noise looks smaller than it is (2.4 shows how much).

**2. Profiling out the clone frequencies.** Every clone has its own unknown
$\lambda_c$ (thousands of nuisance parameters). For each trial value of $\phi$, every
$\lambda_c$ is set to its maximum-likelihood value (Fisher scoring), which gives the
profile likelihood $\ell_p(\phi) = \max_\lambda \ell(\phi, \lambda)$.

**3. Cox–Reid adjustment.** Each clone has only 2–5 observations but its own fitted
$\hat\lambda_c$, which absorbs part of the noise, the same problem as dividing a
variance by $n$ instead of $n-1$. The adjusted profile likelihood

$$\ell_{CR}(\phi) = \ell_p(\phi) - \tfrac12 \sum_c \log I_c(\hat\lambda_c, \phi)$$

($I_c$: the Fisher information for $\log\lambda_c$) removes that bias. It is the
standard estimator of edgeR (McCarthy, Chen & Smyth 2012).

**Clones used.** Detected in at least 2 samples (a clone seen once carries no
information about variation between samples), with frequency at least 10× the
detection cutoff ($\ge 10^{-5}$) over its detected samples. Every clone is still
*tested* later; this filter only concerns estimation.

The table compares, on the real donors, the chosen estimator with two simplified ones.

```python
def phi_global(g, truncate=True, cox_reid=True):
    return nb.fit_nb_dispersion_cr(g, min_detect=2, min_freq_mult=10, truncate=truncate,
                                   c_col='c_t', cox_reid=cox_reid, n_bins=1,
                                   min_bin_clones=10**9)['phi']

est = pd.DataFrame({short(p): {
    'truncated + Cox-Reid (used)': phi_global(g),
    'not truncated': phi_global(g, truncate=False),
    'no Cox-Reid': phi_global(g, cox_reid=False)} for p, g in d_qc.groupby('patient')}).T
display(est.round(3))
```

The three estimators disagree substantially, so the choice matters. The next
subsection decides it by simulation: which one recovers a $\phi$ we know?

## 2.4 Simulation study: does the estimator recover a known $\phi$?

Simulated data reproduce each donor's **real** design: the same sampled days, depths,
offsets and cutoffs, the same clone frequencies (each real clone's frequency, so the
power-law clone-size distribution of Section 1.3 is kept), and the same cell-sampling
dropout (Section 3). Counts are drawn from the NB with a chosen $\phi$; only counts at
or above the cutoff are kept, exactly as in the real data. The estimators are then
applied to the simulated data.

Two scenarios:

* **neutral**: every clone has a constant frequency (the estimator's assumption);
* **5 % changing clones**: 5 % of the clones follow a log-linear trend of 2–30-fold over
  the series. The estimator assumes constant frequencies, so it can only read these
  changes as noise; the question is in which direction, and by how much, $\phi$ moves.

```python
SIM_DONORS = ['HD106_SEQTR', 'HDBR03_SEQTR_ReSeq', 'HDBR06_SEQTR_ReSeq']
SIM_PHI = [1.0, 1.6, 3.0]

rows = []
for p in SIM_DONORS:
    g = d_qc[d_qc['patient'] == p]
    for phi_true in SIM_PHI:
        Nt = nb.estimate_dropout_N(g, phi=phi_true)
        for scen, frac in [('neutral', 0.0), ('5 % changing clones', 0.05)]:
            sim, _ = nb.simulate_like_patient(g, phi_true, Nt, frac_spike=frac, seed=11)
            for name, kw in [('truncated + Cox-Reid (used)', {}),
                             ('not truncated', dict(truncate=False)),
                             ('no Cox-Reid', dict(cox_reid=False))]:
                if scen != 'neutral' and name != 'truncated + Cox-Reid (used)':
                    continue
                rows.append({'donor design': short(p), 'phi_true': phi_true,
                             'scenario': scen, 'estimator': name,
                             'phi_hat': phi_global(sim, **kw)})
simres = pd.DataFrame(rows)
simres['phi_hat / phi_true'] = simres['phi_hat'] / simres['phi_true']
display(simres.pivot_table(index=['scenario', 'estimator'], columns='phi_true',
                           values='phi_hat / phi_true', aggfunc=['min', 'max']).round(2))
```

```python
fig, ax = plt.subplots(figsize=(7.2, 3.8))
styles = {('neutral', 'truncated + Cox-Reid (used)'): ('#2a78d6', 'o'),
          ('neutral', 'not truncated'): ('#eb6834', 's'),
          ('neutral', 'no Cox-Reid'): ('#8a8984', '^'),
          ('5 % changing clones', 'truncated + Cox-Reid (used)'): ('#1baf7a', 'D')}
for k, ((scen, name), (col, mk)) in enumerate(styles.items()):
    s = simres[(simres['scenario'] == scen) & (simres['estimator'] == name)]
    x = np.log(s['phi_true']) + (k - 1.5) * 0.045
    ax.scatter(x, s['phi_hat / phi_true'], color=col, marker=mk, s=34, edgecolor='white',
               linewidth=0.6, zorder=3, label=f'{name} — {scen}')
ax.axhline(1, color=INK, lw=1)
ax.axhspan(0.9, 1.1, color=GREY, alpha=0.12, lw=0)
ax.set(xticks=np.log(SIM_PHI), xticklabels=[str(v) for v in SIM_PHI], xlabel='true φ',
       ylabel='estimated / true φ', yscale='log',
       title='Recovery of a known φ (three donor designs per point group)')
ax.legend(fontsize=7.5, loc='upper left', bbox_to_anchor=(1.01, 1))
fig.tight_layout()
```

**What this shows.**

* **The Cox–Reid adjustment is essential.** Without it, $\phi$ is overestimated by
  40–80 %: the noise would be understated and ordinary fluctuations called.
* **Truncation matters at the donors' values of $\phi$.** Ignoring it overestimates
  $\phi$ when $\phi$ is small (the counts that fell below the cutoff are missing from the
  low tail): by 27–35 % at $\phi = 1$, 9–14 % at $\phi = 1.6$, and only 2–3 % at
  $\phi = 3$. This is the anti-conservative direction.
* **The truncated, Cox–Reid estimator recovers $\phi$ to within 1 % at $\phi = 3$ and
  4–6 % at $\phi = 1.6$, and underestimates it by 12–19 % at $\phi = 1$**, the
  conservative direction: the null is, if anything, noisier than the truth.
* **Changing clones push $\phi$ down** (by 11–28 % with 5 % changing clones). Treating
  real change as noise makes the data look noisier, so $\phi$ falls: again the
  conservative direction. Healthy donors are expected to have little change, so the
  effect on their estimates should be small.

The truncated, Cox–Reid estimator is the one used from here on.

## 2.5 Is one $\phi$ enough? Frequency-dependent dispersion

Section 2.2 hinted that abundant clones are relatively less noisy. To test it, each
donor's clones are split into six bins of frequency (equal numbers of clones) and $\phi$
is estimated separately in each bin, with the same estimator.

This is a well-documented feature of sequencing counts: DESeq2 and edgeR model the
dispersion as a function of mean abundance (Love, Huber & Anders 2014; McCarthy, Chen &
Smyth 2012), and TCR noise models use a variance $m + a\,m^\gamma$ with $\gamma < 2$
(Puelma Touzel, Walczak & Mora 2020). A plausible source here is cell sampling: a rare
clone is represented by few cells in a sample, which adds noise that fades as the
clone grows (Section 3).

```python
trends = nb.fit_phi_trends(d_qc)

fig, ax = plt.subplots(figsize=(6.6, 4.2))
ff = np.linspace(-5, -3.8, 50)
for p in DONORS:
    r = trends[p]
    b = r['bins']
    ok = ~b['at_bound']
    ax.scatter(10 ** b.loc[ok, 'log10_freq'], b.loc[ok, 'phi'], color=COLOR[p], s=30,
               edgecolor='white', linewidth=0.6, zorder=3)
    ax.plot(10 ** ff, r['trend'](10 ** ff), color=COLOR[p], lw=1.4, label=short(p))
ax.set(xscale='log', yscale='log', xlabel='clone frequency',
       ylabel='φ (estimated in the bin)',
       title='Dispersion rises with clone frequency (points: bins, lines: fitted trend)')
ax.legend(fontsize=7.5, loc='upper left', bbox_to_anchor=(1.01, 1))
tidy_log(fig)
fig.tight_layout()

tab = pd.DataFrame({short(p): {'phi (one value)': r['phi_global'],
                               'slope b: d log φ / d log10 f': r['b'],
                               'φ at 1e-5': float(r['trend'](1e-5)),
                               'φ at 1e-4': float(r['trend'](1e-4)),
                               'bins at search bound': int(r['bins']['at_bound'].sum())}
                    for p, r in trends.items()}).T
display(tab.round(2))
```

Every donor shows the same pattern: $\phi$ rises with frequency, by a factor of about
2–4 between $10^{-5}$ and $10^{-4}$. The trend is a line in $(\log_{10} f,\ \log\phi)$,
fitted to the bins with weights proportional to their size, used from the estimation
floor ($10^{-5}$) to the highest bin and held constant beyond (no extrapolation). A bin
whose fit runs into the search bound is left out (HDBR06's rarest bin).

**Is the trend real or an artifact of the estimator?** Two checks by simulation:

1. **Constant $\phi$, every donor's design.** Each donor is simulated with one $\phi$
   for all clones (its own single-value estimate) and binned exactly like the real data.
   Any slope that appears is produced by the estimator, not by the data.
2. **A known trend** (HDBR06's own fitted trend, on HDBR06's design): is it recovered?

```python
rows = []
fig, axes = plt.subplots(2, 3, figsize=(11, 6.2), sharex=True)
for ax, p in zip(axes.ravel(), DONORS):
    g = d_qc[d_qc['patient'] == p]
    phi_c = trends[p]['phi_global']
    sim, _ = nb.simulate_like_patient(g, phi_c, nb.estimate_dropout_N(g, phi=phi_c),
                                      frac_spike=0.0, seed=1)
    r_sim = nb.fit_phi_trend(sim)
    rb = trends[p]['bins'][~trends[p]['bins']['at_bound']]
    sb = r_sim['bins'][~r_sim['bins']['at_bound']]
    ax.plot(10 ** rb['log10_freq'], rb['phi'], 's-', color=COLOR[p], ms=4.5, label='real')
    ax.plot(10 ** sb['log10_freq'], sb['phi'], 'o--', color=GREY, ms=4, lw=1.2,
            label='simulated, constant φ')
    ax.axhline(phi_c, color=INK, lw=0.9, ls=':', label='φ used in the simulation')
    ax.set(xscale='log', yscale='log', title=short(p))
    # bias-corrected real bins: real phi x (true phi / phi the estimator returns at that frequency)
    corr = rb['phi'].to_numpy() * phi_c / np.interp(rb['log10_freq'], sb['log10_freq'], sb['phi'])
    rows.append({'donor': short(p), 'slope b, real': trends[p]['b'],
                 'slope b, constant-φ simulation': r_sim['b'],
                 'rise across bins, real': rb['phi'].iloc[-1] / rb['phi'].iloc[0],
                 'rise across bins, bias-corrected': corr[-1] / corr[0]})
for ax in axes[-1]:
    ax.set_xlabel('clone frequency')
for ax in axes[:, 0]:
    ax.set_ylabel('φ in bin')
axes[0, 0].legend(fontsize=7.5, loc='upper left')
fig.suptitle('1. Real bins vs. a simulation with constant φ on the same design', y=1.01)
tidy_log(fig)
fig.tight_layout()
display(pd.DataFrame(rows).set_index('donor').round(2))
```

```python
# 2. a known trend, on HDBR06's design
g06 = d_qc[d_qc['patient'] == 'HDBR06_SEQTR_ReSeq']
true_trend = trends['HDBR06_SEQTR_ReSeq']['trend']
Nt_t = nb.estimate_dropout_N(g06, phi=true_trend)
fig, ax = plt.subplots(figsize=(6.0, 3.8))
ax.plot(10 ** ff, true_trend(10 ** ff), color=INK, lw=2, label='true trend')
for seed in (1, 2):
    sim, _ = nb.simulate_like_patient(g06, true_trend, Nt_t, frac_spike=0.0, seed=seed)
    r = nb.fit_phi_trend(sim)
    ax.plot(10 ** ff, r['trend'](10 ** ff), color='#e87ba4', lw=1.4,
            label='estimated trend (2 simulations)' if seed == 1 else None)
ax.set(xscale='log', yscale='log', xlabel='clone frequency', ylabel='φ',
       title='2. A known trend (HDBR06) is recovered below the truth')
ax.legend(fontsize=7.5)
tidy_log(fig)
fig.tight_layout()
```

**What this shows.**

1. **Part of the slope is the estimator's own bias, but not most of it.** With a constant
   $\phi$ the bins still rise slightly (slope $b$ 0–0.5, against 0.9–1.5 in the real
   data): in the rarest bins the estimator underestimates $\phi$, more so when $\phi$ is
   small (the same bias as at $\phi = 1$ in 2.4). In the most abundant bins it returns
   the true value. The real bins rise far above the constant-$\phi$ simulation at high
   frequency in every donor, and after correcting each real bin for the bias seen in the
   simulation, $\phi$ still rises by about 1.5–3.5× from the rarest to the most abundant
   bin. The frequency dependence is real.
2. **A known trend is recovered below the truth everywhere**: about 15 % low at the
   abundant end and about half the true value at the floor ($10^{-5}$).

The fitted trend is used as it is, without a bias correction. Its error is one-sided: it
understates $\phi$ for rare clones, i.e. treats them as noisier than they are, which is
the conservative direction. For abundant clones, where the trend gives the test its
extra power, it is close to unbiased.

The test therefore uses a **frequency-dependent dispersion** $\phi(f)$: each clone gets
the $\phi$ that corresponds to its frequency. Compared with a single $\phi$ fitted
mostly on rare clones, this treats rare clones as noisier and abundant clones as less
noisy, which is what the donors show.

## 2.6 The reference dispersion for patients, $\phi_\text{ref}(f)$

Patients have no healthy baseline of their own, so their null comes entirely from the
donors. To be conservative, each frequency takes the **noisiest donor** at that
frequency:

$$\phi_\text{ref}(f) = \min_{d \in \text{donors}} \phi_d(f).$$

A patient clone of frequency $f$ is therefore never treated as less noisy than the same
clone would be in the noisiest healthy donor. (The test is applied to clones with
average frequency $\ge 10^{-5}$, the floor of the curve; Section 4.)

Two donors set the curve: HDBR06 below about $6 \times 10^{-5}$ and HDBR01 above. At the
floor, $\phi_\text{ref}$ is also where the estimator's downward bias (2.5) is largest, so
the rarest clones get the most cautious null.

```python
phi_ref = nb.reference_phi_trend(trends)

fig, ax = plt.subplots(figsize=(6.6, 4.0))
for p in DONORS:
    ax.plot(10 ** ff, trends[p]['trend'](10 ** ff), color=COLOR[p], lw=1.1, alpha=0.8,
            label=short(p))
ax.plot(10 ** ff, phi_ref(10 ** ff), color=INK, lw=3, label='φ_ref(f) = minimum')
ax.set(xscale='log', yscale='log', xlabel='clone frequency', ylabel='φ',
       title='Reference dispersion: the noisiest donor at each frequency')
ax.legend(fontsize=7.5, loc='upper left', bbox_to_anchor=(1.01, 1))
tidy_log(fig)
fig.tight_layout()

grid = np.array([1e-5, 2e-5, 5e-5, 1e-4, 2e-4])
noisiest = [short(DONORS[int(np.argmin([trends[p]['trend'](f) for p in DONORS]))]) for f in grid]
display(pd.DataFrame({'frequency': grid, 'φ_ref': phi_ref(grid).round(2),
                      'noisiest donor': noisiest}))
```

## 2.7 Leave-one-donor-out

Two questions, each answered by leaving every donor out in turn.

**a. How much does $\phi_\text{ref}(f)$ depend on any single donor?** If one donor
dominated the reference, removing it would move the curve a lot.

**b. Is the null calibrated?** Each donor is tested *as if it were a patient*, with the
reference built from the **other** donors only. Healthy donors are, by assumption,
neutral, so the p-values should be uniform (or conservative): about 5 % below 0.05 and
no clones called after multiple-testing correction. The test is the full pipeline test
(slope likelihood-ratio test with censoring and cell-sampling dropout, clones with
frequency $\ge 10^{-5}$), explained in Sections 3 and 4; here it serves to check the null.
The same check with a single, constant $\phi$ (the noisiest donor's overall value) is
shown for comparison.

```python
fig, ax = plt.subplots(figsize=(6.6, 4.0))
ax.plot(10 ** ff, phi_ref(10 ** ff), color=INK, lw=3, label='all donors')
for p in DONORS:
    ref_wo = nb.reference_phi_trend(trends, exclude=p)
    ax.plot(10 ** ff, ref_wo(10 ** ff), color=COLOR[p], lw=1.3, ls='--',
            label=f'without {short(p)}')
ax.set(xscale='log', yscale='log', xlabel='clone frequency', ylabel='φ_ref',
       title='a. Reference curve with each donor left out')
ax.legend(fontsize=7.5, loc='upper left', bbox_to_anchor=(1.01, 1))
tidy_log(fig)
fig.tight_layout()
```

```python
RUN_CONSTANT = True          # also run the comparison with one constant phi
loo = []
for held, g in d_qc.groupby('patient'):
    specs = {'φ_ref(f), trend': nb.reference_phi_trend(trends, exclude=held)}
    if RUN_CONSTANT:
        specs['constant φ'] = min(r['phi_global'] for p, r in trends.items() if p != held)
    for name, phi_spec in specs.items():
        Nt = nb.estimate_dropout_N(g, phi=phi_spec)
        r = nb.nb_trajectory_test(g, phi_spec, likelihood='dropout', dropout_N=Nt,
                                  min_freq=1e-5, beta_bound=np.log(100), n_jobs=N_JOBS)
        r = nb.add_qvalues(r, 'p_value', by='patient')
        loo.append(r.assign(held_out=short(held), null=name))
loo = pd.concat(loo, ignore_index=True)

calib = loo.groupby(['held_out', 'null']).agg(
    tested=('clono', 'size'),
    FPR_05=('p_value', lambda p: (p < .05).mean()),
    FPR_01=('p_value', lambda p: (p < .01).mean()),
    BH_calls=('q_value', lambda q: int((q < .05).sum()))).unstack('null')
display(calib.round(4))
```

```python
# false-positive rate by clone frequency (average over the person's samples)
Dsum = d_qc.groupby(['patient', 'day'])['D_t'].first().groupby('patient').sum()
fq = d_qc.groupby(['patient', 'clono'], as_index=False)['count'].sum()
fq['freq'] = fq['count'] / fq['patient'].map(Dsum)
fq['held_out'] = fq['patient'].map(short)
loo_f = loo.merge(fq[['held_out', 'clono', 'freq']], on=['held_out', 'clono'])
loo_f['frequency band'] = pd.cut(loo_f['freq'], [1e-5, 2e-5, 5e-5, 2e-4, 1],
                                 labels=['1–2e-5', '2–5e-5', '5e-5–2e-4', '> 2e-4'])

fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
for ax, (lab, sub) in zip(axes, [('all donors', loo_f),
                                 ('without HDBR03', loo_f[loo_f['held_out'] != 'HDBR03'])]):
    t = (sub.groupby(['frequency band', 'null'], observed=True)['p_value']
            .apply(lambda p: (p < .05).mean()).unstack())
    x = np.arange(len(t))
    for k, (name, col) in enumerate(zip(t.columns, [GREY, '#2a78d6'])):
        ax.bar(x + (k - 0.5) * 0.36, t[name], width=0.34, color=col, label=name)
    ax.axhline(0.05, color='#d03b3b', lw=1, ls='--', label='nominal 5 %')
    ax.set(xticks=x, xticklabels=t.index, ylabel='fraction with p < 0.05',
           title=f'b. False-positive rate by clone frequency ({lab})', ylim=(0, 0.1))
    ax.grid(axis='x', visible=False)
axes[0].legend(fontsize=8)
fig.tight_layout()
```

```python
# p-value distributions (should be uniform or tilted towards 1)
fig, axes = plt.subplots(1, len(DONORS), figsize=(13, 2.6), sharey=True)
for ax, p in zip(axes, DONORS):
    s = loo[(loo['held_out'] == short(p)) & (loo['null'] == 'φ_ref(f), trend')]['p_value']
    ax.hist(s, bins=20, range=(0, 1), color=COLOR[p], density=True)
    ax.axhline(1, color=INK, lw=1, ls='--')
    ax.set(title=short(p), xlabel='p-value')
    ax.grid(axis='x', visible=False)
axes[0].set_ylabel('density')
fig.suptitle('p-values of each held-out donor under φ_ref(f) from the other donors', y=1.04)
fig.tight_layout()
```

**What this shows.**

* **a.** The reference curve is set by HDBR06 (rare clones) and HDBR01 (abundant clones).
  Removing HDBR06 raises the low end by about 15 %, to HDBR01's curve; removing HDBR01
  raises the plateau by about 7 %, to HDBR06's. Removing any other donor changes nothing.
  No single donor determines the null.
* **b.** For five of the six donors the false-positive rate is 0.1–3.2 % (nominal 5 %),
  below 5 % in every frequency band, and no clone is called after multiple-testing
  correction. The p-value histograms are flat (HDBR01) or tilted towards 1: the null is
  calibrated, and conservative.
* **Compared with a constant $\phi$**, the frequency-dependent null moves the caution
  to where the noise is: the false-positive rate falls for the rarest clones and rises for
  the more abundant ones, which a single $\phi$ (fitted mostly on rare clones) treated as
  noisier than they are. It stays below nominal in every band. This is where the extra
  power for abundant clones comes from.
* **HDBR03 is the exception**, as in every earlier analysis: 8 % of its clones have
  $p < 0.05$, with 34 clones called after correction (6 with a constant $\phi$). The cell
  below shows what they are: clones present at high counts in the early samples and
  absent from the late ones, or the reverse, mostly at frequencies $2 \times 10^{-5}$ to
  $2 \times 10^{-4}$, where the trended null is less cautious than a constant one. These
  are changes carried by groups of samples, not by one noisy point, which is the kind of
  change the test is built to find. Whether they are real turnover in a healthy person or
  a sample-level difference between the early and late samples cannot be decided from
  the counts. Either way they are a reminder that "healthy" does not guarantee "static",
  and the reason the reference takes the noisiest donors rather than an average.

```python
# HDBR03: the clones called under phi_ref(f), and the samples they were detected in
c03 = loo[(loo['held_out'] == 'HDBR03') & (loo['null'] == 'φ_ref(f), trend')
          & (loo['q_value'] < .05)]
g03 = d_qc[d_qc['patient'] == 'HDBR03_SEQTR_ReSeq']
days03 = np.sort(g03['day'].unique())
cnt03 = (g03.pivot_table(index='clono', columns='day', values='count', aggfunc='sum')
            .reindex(columns=days03).loc[c03['clono']])
pattern = cnt03.notna().astype(int).astype(str).agg(''.join, axis=1)
tab03 = pd.DataFrame({'direction': np.where(c03['beta'].to_numpy() > 0, 'expansion',
                                             'contraction'),
                      'detected in days ' + ' '.join(map(str, days03)): pattern.to_numpy()})
display(pd.crosstab(tab03.iloc[:, 1], tab03['direction']))
display(cnt03.assign(beta=c03['beta'].to_numpy()).sort_values('beta').head(12))
```

## 2.8 Summary

| Observation | Decision |
|---|---|
| Variance far above Poisson, growing with $\mu^2$ (2.2) | Negative binomial with offsets $L_t = D_t \times \text{TMM}$ |
| Without Cox–Reid, $\phi$ overestimated by 40–80 % (2.4) | Cox–Reid adjusted profile likelihood |
| Ignoring truncation overestimates $\phi$ at the donors' values (2.4) | Likelihood truncated at the cutoff $c_t$ |
| The chosen estimator recovers $\phi$, erring conservative; changing clones lower $\phi$ (2.4) | Estimator validated; healthy-donor fits are reliable |
| $\phi$ rises with frequency in every donor, 1.5–3.5× after correcting the estimator's own bias (2.5) | Frequency-dependent dispersion $\phi(f)$ |
| Patients have no healthy baseline | $\phi_\text{ref}(f)$ = noisiest donor at each frequency (2.6) |
| Leave-one-donor-out: stable reference, calibrated test except HDBR03 (2.7) | Donor-based null validated |
