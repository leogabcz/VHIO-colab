```python
nb_test = fit_nb_dispersion_trended_smooth(healthy_df[healthy_df["patient"] == "HD106_SEQTR"])
```

```python
nb_test2 = fit_nb_dispersion_trended_smooth(healthy_df[healthy_df["patient"] == "HDBR02_SEQTR_ReSeq"])
nb_test3 = fit_nb_dispersion_trended_smooth(healthy_df[healthy_df["patient"] == "HD197_SEQTR"])
nb_test4 = fit_nb_dispersion_trended_smooth(healthy_df[healthy_df["patient"] == "HDBR06_SEQTR_ReSeq"])
```

```python
nb_pooled = pool_dispersion_trends_equal_weight([nb_test, nb_test2, nb_test3, nb_test4])
```

```python

plt.plot(nb_test['smooth_x'], nb_test['smooth_log_phi'], alpha=0.4, label='P1')
plt.plot(nb_test2['smooth_x'], nb_test2['smooth_log_phi'], alpha=0.4, label='P2')
plt.plot(nb_test3['smooth_x'], nb_test3['smooth_log_phi'], alpha=0.4, label='P3')
plt.plot(nb_test4['smooth_x'], nb_test4['smooth_log_phi'], alpha=0.4, label='P4')
plt.plot(nb_pooled['smooth_x'], nb_pooled['smooth_log_phi'], color='black', lw=2, label='pooled (equal-weight)')
plt.xlabel('log10(mean frequency)'); plt.ylabel('log(phi)'); plt.legend()
plt.savefig("nb_test.png")
# and the spread diagnostic
plt.figure()
plt.plot(nb_pooled['smooth_x'],nb_pooled['per_donor_log_phi'].std(axis=0))
plt.xlabel('log10(mean frequency)'); 
plt.ylabel('std across donors (log phi)')
plt.savefig("std.png")
```
